```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl

# =============================================================================
# IEEE CONFERENCE MATPLOTLIB SETTINGS
# =============================================================================
mpl.rcParams.update({
    'font.family': 'serif',
    'font.serif': ['Times New Roman'],
    'mathtext.fontset': 'stix',
    'axes.labelsize': 9,                
    'font.size': 9,                     
    'legend.fontsize': 8,
    'xtick.labelsize': 8,
    'ytick.labelsize': 8,
    'xtick.direction': 'in',            
    'ytick.direction': 'in',
    'xtick.top': True,                  
    'ytick.right': True,                
    'axes.linewidth': 2.0,              
    'lines.linewidth': 1.5,
    'figure.dpi': 300,
    'savefig.dpi': 300,
    'savefig.bbox': 'tight',
    'savefig.pad_inches': 0.05
})

def plot_milt_vs_mipt_phase_diagram(save_path="Phase_Diagram_MILT_vs_MIPT_IEEE.pdf"):
    # =========================================================================
    # HARDCODED DATA
    # =========================================================================
    # MILT Data (HEA2)
    s_milt = np.array([1.0, 0.9, 0.8, 0.7, 0.6, 0.5, 0.4, 0.3])
    pc_milt = np.array([0.2218, 0.2240, 0.2263, 0.2549, 0.2899, 0.3615, 0.4955, 0.7833])

    # MIPT 1 Data (HEA2)
    s_mipt = np.array([1.0, 0.9, 0.8, 0.7, 0.6, 0.5, 0.4])
    pc_mipt = np.array([0.3897, 0.4741, 0.4991, 0.4938, 0.6047, 0.7491, 0.9347])
    err_mipt = np.array([0.0001, 0.0084, 0.0017, 0.0081, 0.0005, 0.0019, 0.0004])

    # =========================================================================
    # PLOTTING
    # =========================================================================
    fig, ax = plt.subplots(figsize=(3.5, 2.5))

    # Plot MILT Boundary
    ax.plot(
        pc_milt, s_milt, 
        marker='s', linestyle='-', color='#21918c', # Teal
        markerfacecolor='none', markersize=4.0, markeredgewidth=1.2, 
        label="MILT Boundary"
    )

    # Plot MIPT Boundary (with error bars)
    ax.errorbar(
        pc_mipt, s_mipt, xerr=err_mipt,
        marker='o', linestyle='--', color='#440154', # Dark Indigo
        markerfacecolor='none', markersize=4.0, markeredgewidth=1.2,
        capsize=2, capthick=1.0, label="MIPT Boundary"
    )

    # =========================================================================
    # FORMATTING
    # =========================================================================
    ax.set_xlabel("Critical Measurement Rate, $p_c$")
    ax.set_ylabel("Measurement Strength, $s$")
    
    # Set limits to span exactly 0 to 1 with a tiny margin for visual breathing room
    ax.set_xlim(0.0, 1.05)
    ax.set_ylim(0.0, 1.05)
    
    # Tick mark styling
    ax.tick_params(axis='both', which='major', width=1.5, length=4.0)
    ax.grid(False)

    # Clean Legend
    ax.legend(
        loc="upper right", 
        frameon=True, 
        edgecolor='black', 
        fancybox=False, 
        framealpha=1.0,
        handlelength=1.5,
        handletextpad=0.5
    )

    # Add text label for the Ansatz to keep it self-contained
    ax.text(0.05, 0.05, "Ansatz: HEA2", transform=ax.transAxes, 
            fontsize=8, verticalalignment='bottom', 
            bbox=dict(boxstyle='round,pad=0.3', facecolor='white', edgecolor='gray', alpha=0.9))

    plt.tight_layout()
    
    # =========================================================================
    # SAVE AND SHOW
    # =========================================================================
    if save_path:
        plt.savefig(save_path)
        print(f"Figure saved successfully to: {save_path}")
        
    plt.show()

if __name__ == "__main__":
    plot_milt_vs_mipt_phase_diagram()
```
